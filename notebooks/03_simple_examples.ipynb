# Simple Examples and Use Cases

This notebook provides simple, quick examples of using Magrathea for common tasks.

```python
import magrathea as mag
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = (10, 6)
```

## Example 1: Quick Earth Model

```python
# Create Earth model in just a few lines
earth = mag.Planet()
earth.add_layer('core', mass=0.33)
earth.add_layer('mantle', mass=0.67)
earth.surface_temp = 300

print("Created Earth-like planet:")
print(earth)
```

## Example 2: Super-Earth

```python
# Model a 5 Earth-mass super-Earth
super_earth = mag.Planet()
super_earth.add_layer('core', mass=1.65)  # 33% of 5 M_Earth
super_earth.add_layer('mantle', mass=3.35)  # 67% of 5 M_Earth
super_earth.surface_temp = 400  # Hotter due to closer orbit

print(f"Super-Earth model: {super_earth.total_mass:.1f} Earth masses")

try:
    super_earth.solve()
    print(f"Predicted radius: {super_earth.results.radius:.3f} Earth radii")
except Exception as e:
    print(f"Note: Solve requires compiled Magrathea executable")
```

## Example 3: Water World

```python
# Create a water-rich planet
water_world = mag.Planet()
water_world.add_layer('core', mass=0.2)
water_world.add_layer('mantle', mass=0.3)
water_world.add_layer('hydro', mass=0.5)  # 50% water!
water_world.surface_temp = 300

print("Water World composition:")
for layer in water_world.layers:
    print(f"  {layer.name}: {layer.mass:.2f} M_Earth ({100*layer.mass/water_world.total_mass:.1f}%)")
```

## Example 4: Using Different Phase Diagrams

```python
# Check available phase diagrams
print("Available phase diagrams:")
print("\nCore:")
for name, desc in mag.eos.CORE_PHASE_DIAGRAMS.items():
    print(f"  {name}: {desc}")

print("\nMantle:")
for name, desc in mag.eos.MANTLE_PHASE_DIAGRAMS.items():
    print(f"  {name}: {desc}")

# Use alternative phase diagram
planet = mag.Planet()
planet.add_layer('core', mass=0.33, phase_diagram='Fe_default')
planet.add_layer('mantle', mass=0.67, phase_diagram='Si_default')
print(f"\nCreated planet with custom EOS choices")
```

## Example 5: Comparing Planets Side-by-Side

```python
# Create several planets
planets = []
labels = []

# Pure rock
p1 = mag.Planet()
p1.add_layer('core', mass=0.33)
p1.add_layer('mantle', mass=0.67)
p1.surface_temp = 300
planets.append(p1)
labels.append('Rocky')

# With water
p2 = mag.Planet()
p2.add_layer('core', mass=0.2)
p2.add_layer('mantle', mass=0.5)
p2.add_layer('hydro', mass=0.3)
p2.surface_temp = 300
planets.append(p2)
labels.append('Water-rich')

# Larger core
p3 = mag.Planet()
p3.add_layer('core', mass=0.6)
p3.add_layer('mantle', mass=0.4)
p3.surface_temp = 300
planets.append(p3)
labels.append('Iron-rich')

print("Planet Comparison:")
print(f"{'Type':<15} {'Core':<10} {'Mantle':<10} {'Water':<10}")
print("-" * 50)
for i, planet in enumerate(planets):
    core = planet.get_layer('core')
    mantle = planet.get_layer('mantle')
    water = planet.get_layer('hydro')
    
    core_mass = core.mass if core else 0
    mantle_mass = mantle.mass if mantle else 0
    water_mass = water.mass if water else 0
    
    print(f"{labels[i]:<15} {core_mass:<10.2f} {mantle_mass:<10.2f} {water_mass:<10.2f}")
```

## Example 6: Unit Conversions

```python
# Magrathea includes useful unit conversion functions
print("Unit Conversions:")
print(f"\n1 Earth mass = {mag.units.M_earth:.3e} kg")
print(f"1 Earth radius = {mag.units.R_earth:.3e} m")
print(f"\nJupiter mass = {mag.units.jupiter_masses_to_earth_masses(1):.1f} Earth masses")
print(f"\nPressure units:")
print(f"  1 GPa = {mag.units.GPa:.2e} Pa")
print(f"  1 bar = {mag.units.bar:.2e} Pa")
```

## Example 7: Create Configuration File Manually

Advanced users can create configuration files directly.

```python
from magrathea import Config, RunMode

# Create a configuration
config = Config(RunMode.FULL_SOLVER)
config.set('mass_of_core', 0.33)
config.set('mass_of_mantle', 0.67)
config.set('surface_temp', 300)
config.set('output_file', './result/my_planet.txt')

# View the configuration
print("Generated Configuration:")
print("-" * 60)
print(config)
print("-" * 60)

# Save to file (optional)
# config.save('my_planet.cfg')
```
